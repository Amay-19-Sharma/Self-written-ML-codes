```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.stats import t
```

```python
#(one sample testing) simple t- test
def t_score(sample_mean,sample_std,pop_mean,sample_size):
    std_error = sample_std/sample_size**0.5 
    t_value = abs(sample_mean-pop_mean)/std_error
    return t_value
```

```python
sample = [35, 40, 38, 42, 30, 50, 25, 45, 33, 37, 36, 39, 41, 34, 31, 48, 22, 55, 32, 37]
sample = np.array(sample)
```

```python
sample_mean = sample.mean()
pop_mean = 40
sample_std = sample.std()
sample_size = 20
t_value = t_score(sample_mean,sample_std,pop_mean,sample_size)
t_value
```

```python
#degree of freedom = sample_size - 1
#now find t-critical value - depends only on alpha and degrees of freedom
from scipy.stats import t
df = sample_size - 1
alpha = 0.05
t_critical = abs(t.ppf(alpha,df))
t_critical
```

```python
if t_value>t_critical:
    print("We have sufficient evidence to reject null hypothesis.")
else: 
    print("We failed to reject null hypothesis")
```

```python
#(2groups test (independent))
sample1 = [35, 38, 36, 32, 40, 37, 34, 39, 33, 36,34,34,38,31,32]
sample2 = [32, 36, 34, 30, 38, 33, 35, 37, 31, 34]
#function here
def t_score(sample1_mean,sample1_std,sample2_mean,sample2_std,sample1_size,sample2_size):
    std_error = sample1_std**2/sample1_size + sample2_std**2/sample2_size
    t_value = abs(sample1_mean-sample2_mean)/std_error**0.5
    return t_value
```

```python
sample1 = np.array(sample1)
sample2 = np.array(sample2)
sample1_mean = sample1.mean()
sample1_std = sample1.std()
sample2_mean = sample2.mean()
sample2_std = sample1.std()
sample1_size = len(sample1)
sample2_size = len(sample2)
t_value = t_score(sample1_mean,sample1_std,sample2_mean,sample2_std,sample1_size,sample2_size)
t_value
```

```python
deg_f = sample1_size + sample2_size - 2
t_cric = abs(t.ppf(alpha,deg_f))
t_cric
```

```python
if t_value>t_cric:
    print("We have sufficient evidence to reject null hypothesis.")
else: 
    print("We failed to reject null hypothesis")
```

```python
import numpy as np
from scipy.stats import t

before = np.array([70, 65, 80, 75, 85, 90, 60, 55, 70, 75])
after  = np.array([75, 68, 82, 78, 88, 95, 62, 58, 80, 68])
```

```python
#PAIRED HYPOTHESIS TESTING
def t_score(diff_mean,diff_std,sample_size):
  std_error = abs(diff_std)/sample_size**0.5
  t_value = abs(diff_mean)/std_error
  return t_value

sample_size = len(before)
before = np.array(before)
after = np.array(after)
diff = np.subtract(before,after)
diff_mean = diff.mean()
diff_std = diff.std()
#calculate Degrees of freedom
df = sample_size - 1

# Calculate t score
t_value = t_score(diff_mean,diff_std,sample_size)

alpha = 0.05 / 2  # two tailed
t_critical = abs(t.ppf(alpha,df))

# Conclusion
if t_value > t_critical:
    result = "We have sufficient evidence to reject null hypothesis."
else:
    result = "We fail to reject null hypothesis."

print(f"T Value: {t_value}\n"
      f"Critical T Value: {t_critical}\n"
      f"Conclusion: {result}")
```
