```python
#make groups
def make_groups(df,column):
    groups = ['low','average','high','very-high']
    bins = [0,df[column].quantile(0.25),df[column].quantile(0.50),df[column].quantile(0.75),df[column].max()]
    return groups,bins
```

```python
#model evaluation
from sklearn.metrics import mean_squared_error,r2_score
def model_evaluation(MLmodel,x_test,Y_test,name):
    pred = MLmodel.predict(x_test)
    r2 = r2_score(Y_test,pred)
    mse = mean_squared_error(Y_test,pred)
    report = pd.DataFrame({'R2_score':[r2],'MSE':[mse],'Model_Name':[name]})
    return report
```

```python
#OPTIMISING USING GRADIENT DESCENT
#for batch gradient descent we write our own function
#intialising weights (coeff)

def gradient_descent(X1,y1,learning_rate = 0.01,epochs = 1000):
    m = X1.shape[0]  #no of rows
    n = X1.shape[1] #no of features

    #assignment of weight 
    weights = np.zeros(n)

    #cost for storing mse
    cost_history = []
    
    #now we write the iterative step
    for epoch in range(epochs):
        
               #prediction
        prediction = np.dot(X1,weights)
        error = prediction - y1
    
        #assign gradient 
        gradient = (1/m)*(np.dot(X1.T,error))
        #update weights 
        weights -= learning_rate * gradient

        MSE = (1/(2*m))*(np.sum(error**2))
        cost_history.append(MSE)

        #prrint MSE every 100th iteration
        if epoch%100==0:
            print(f'MSE Value at {epoch}th iteration : {MSE}') 
    return weights,cost_history
        
```

```python
#for gradeint descent ->imp points
#always make a seperate feature containg all 1 for the bias terms
#always convert pandas dataframe to numpy array to do mathematical operations

X_train_np = np.hstack(((np.ones((X_train.shape[0],1))),X_train))  #make a column feature containing all ones

#convert y_train to numpy array
y_train_np = y_train.values
```

```python
X_train_np
```

```python
weight,cost_history = gradient_descent(X_train_np,y_train_np,learning_rate=0.01,epochs = 1000)
```

```python
#funtion to plot
def visualise_gd(cost_history1):
    plt.plot(range(len(cost_history1)),cost_history1,marker = 'o',linestyle = '-',color = 'b')
    plt.xlabel('No of Iterations')
    plt.ylabel('Cost Function')
    plt.title('Optimisation')
    plt.grid(True)
    plt.legend()
    plt.show()
visualise_gd(cost_history)
```
